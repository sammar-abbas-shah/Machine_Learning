```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Weather.csv", skiprows=1, encoding='utf-8-sig')
data.columns = data.columns.str.strip() #for spaces issues
X = data.drop(['Play'], axis=1)
Y = data['Play']
```

```python
print("X=", X)
print("Y=", Y)
```

```python
features = list(X)
print(features)
```

```python
prior = {}
likelihoods = {}
pred_priors = {}

for f in features:
    likelihoods[f] = {}
    pred_priors[f] = {}
    for f_val in np.unique(X[f]):
        pred_priors[f].update({f_val: 0})
        for outcome in np.unique(Y):
            likelihoods[f].update({f_val + '' + outcome: 0})
            prior.update({outcome: 0})
```

```python
print("likelihoods=",likelihoods)
####################################################
```
