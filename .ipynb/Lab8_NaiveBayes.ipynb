```python
import numpy as np
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("Weather.csv")
print(data)
```

```python
# now we have to encode categorical data to numerical data

le = LabelEncoder()

# Encode Outlook
data = pd.read_csv("Weather.csv", header=1)
print(data)
label = le.fit_transform(data['Outlook']) 
data.drop('Outlook',axis=1,inplace = True)
data['Outlook']=label                    
print(data)

# Encode Temp
label = le.fit_transform(data['Temp'])
data.drop('Temp',axis=1,inplace = True)
data['Temp']=label
print(data)

# Encode Humidity
label = le.fit_transform(data['Humidity'])
data.drop('Humidity',axis=1,inplace = True)
data['Humidity']=label
print(data)

# Encode Windy
label = le.fit_transform(data['Windy'])
data.drop('Windy',axis=1,inplace = True)
data['Windy']=label
print(data)

# Encode Target Variable (Play)
label = le.fit_transform(data['Play'])
data.drop('Play',axis=1,inplace = True)
data['Play']=label
print(data)
```

```python
#now splitting features and target variable
print(data)
X=data.iloc[:,[0,1,2,3]].values #only pick values
Y=data.iloc[:,[4]].values
Y=np.ndarray.flatten(Y)
```

```python
# now model evaluation using Leave One Out Cross Validation

# Initialize model and cross-validator
model = GaussianNB()
cv= LeaveOneOut()

Y_prediction=[]
Y_actual=[]
```

```python
for train_indx,test_indx in cv.split(X):# 14 times
    X_train = X[train_indx]
    Y_train = Y[train_indx]
    X_test = X[test_indx]
    Y_test = Y[test_indx]
    
    # Train model and predicting the single left out sample
    model.fit(X_train,Y_train)
    Y_prediction.append(model.predict(X_test))
    Y_actual.append(Y_test[0])
# final accuracy 
print("Accruacy of Model = ",accuracy_score(Y_actual,Y_prediction))
```
