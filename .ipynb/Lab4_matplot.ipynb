```python
# Base of the house
x_base = [2, 2, 8, 8, 2]
y_base = [2, 6, 6, 2, 2]

# Roof of the house (triangle)
x_roof = [1, 5, 9, 1]
y_roof = [6, 9, 6, 6]

# Door of the house
x_door = [4, 4, 6, 6, 4]
y_door = [2, 4, 4, 2, 2]

# Plot the shapes
plt.plot(x_base, y_base, c="black")
plt.plot(x_roof, y_roof, c="black")
plt.plot(x_door, y_door, c="black")

# Set axis limits
plt.xlim(0, 10)
plt.ylim(0, 10)

# Hide axes
plt.axis('off')

# Show the plot
plt.show()
```
